#### 3주차 스택과 큐


#### 1. 큐

![nn](img/img_3_02.png)

* 영화관에서 가장 먼저 줄을 선 사람이 먼저 입장하는 것과 동일함
* FIFO(First-In, First-Out) 방식으로 스택과 꺼내는 순서가 반대임
* 큐는 한 쪽 끝은 프런트(front)로 정하여 삭제 연산만 수행함
* 다른 한 쪽 끝은 리어(rear)로 정하여 삽입 연산만 수행함  



#### 1-1. 큐의 연산
* Enqueue: 큐에 데이터를 넣는 작업
* Dequeue: 큐에서 데이터를 꺼내는 작업
* 사이트를 통해서 이해하기 https://visualgo.net/en/list

#### 큐가 많이 사용되는 곳
- 멀티 태스킹을 위한 프로세스 스케쥴링 방식을 구현하기 위해 많이 사용됨

> 프로세스 스케쥴링 방식을 함께 이해하는 것이 좋음

#### 1-2. 자바를 통해 큐를 사용해보기
* Java.util 패키지에서 Queue 관련 클래스를 제공하고 있음
  - Enqueue:  add() 또는 offer() 메서드를 사용함
  - Dequeue: remove() 또는 poll() 메서드를 사용함
  - 자바에서는 스택은 클래스로 구현하여 제공하지만, 큐는 별도의 클래스를 제공하지 있지 않음
  - 따라서 큐 인터페이스를 구현한 클래스들을 사용해야 하며, java.util 패키지에 있는 LinkedList 클래스를 사용해야 함

```java
import java.util.LinkedList; // Queue 사용을 위해, LinkedList 를 import 함
import java.util.Queue;      // Quyeue 사용을 위해 util.Queue를 import 함

// 큐에 들어갈 데이터의 타입을 지정을 위해 파라메터를 넣어야 함
Queue<Integer> ArrayQueue = new LinkedList<Integer>(); // Integer queue 선언
Queue<String> StringQueue = new LinkedList<String>(); // String queue 선언
```

#### 1-3. 큐 구현해보기 

큐를 다루는 enqueue, dequeue 기능 구현해보기 <br>
- JAVA ArrayList 클래스를 활용해야 함
- enqueue와 dequeue 기능을 구현
- dequeue 호출 시, 큐에 데이터가 없다면 , null 을 리턴함 <br>
- 다양한 형태의 데이터를 다루기 위해 Java Genric 활용 

```java
import java.util.ArrayList;

public class DataQueue<D> {
    private ArrayList<D> queue = new ArrayList<D>();
    
    public void enqueue(D value) {
        queue.add(value);
    }
    
    public D dequeue() {
        if (queue.isEmpty()) {
            return null;
        }
        return queue.remove(0);
    }
    
    public static void main(String[] args) {
        DataQueue<Integer> dq = new DataQueue<Integer>();
        dq.enqueue(1);
        dq.enqueue(2);
        dq.enqueue(3);
        System.out.println(dq.dequeue());
        System.out.println(dq.dequeue());
        System.out.println(dq.dequeue());
    }
    
}
```

```java
DataQueue.main(new String[0])
```

#### 2-3. 스택 구현해보기 

- ArrayList로 스택 구현하기 (데이터를 추가하는 것을 push()라고 하고, 꺼내는 것을 pop()라고 함)
- pop을 진행하면서 스택에 데이터가 없다면 null을 리턴함 <br>

```java
import java.util.ArrayList;

public class IntStack<S> {
    private ArrayList<S> stack = new ArrayList<S>();
    
    public void push(S data) {
        stack.add(data);
    }
    
    public S pop() {
        if(stack.isEmpty()) {
            return null;
        }
        return stack.remove(stack.size()-1);
    }
    
    public static void main(String[] args) {
        IntStack<Integer> capacity = new IntStack<Integer>();
        capacity.push(1);
        capacity.push(2);
        capacity.push(3);
        capacity.push(4);
        
        System.out.println(capacity.pop());
        System.out.println(capacity.pop());
        System.out.println(capacity.pop());
        System.out.println(capacity.pop());
    }
}
```

```java
IntStack.main(new String[0])
```
