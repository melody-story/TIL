### 1. 최소 신장 트리 (MST: Minimum Spanning Tree) 알고리즘
- 그래프에서 최소 신장 트리를 찾을 수 있는 알고리즘이 있음
  - Kruskal’s algorithm (크루스칼 알고리즘), Prim's algorithm (프림 알고리즘)

#### 1-2. 크루스칼 알고리즘 (kruskal's algorithm)
1. 모든 정점을 독립적인 집합으로 만듬
2. 모든 간선을 비용을 기준으로 정렬하고, 비용이 작은 간선부터 양 끝의 두 정점을 비교함
3. 두 정점의 최상위 정점을 확인하고, 서로 다를 경우 두 정점을 연결함 
    - 사이클이 생기지 않도록 하는 것임

> 탐욕 알고리즘을 기초로 하고 있음

#### 2.1. Union-Find 구현 

```java
import java.util.HashMap;

HashMap<String, String> parent = new HashMap<String, String>();
HashMap<String, Integer> rank = new HashMap<String, Integer>();
```

```java
public String find(String node) {
    if (parent.get(node) != node) {
        parent.put(node, find(parent.get(node)));
        
    }
    return parent.get(node);
}
```

```java
public void union(String nodeE, String nodeN) {
    String root_1 = find(nodeE);
    String root_2 = find(nodeN);
    
    //union-by-rank
    if(rank.get(root_1) >rank.get(root_2)) {
        parent.put(root_2, root_1);
    } else {
        parent.put(root_1, root_2);
        if (rank.get(root_1) == rank.get(root_2)) {
            rank.put(root_2, rank.get(root_2) +1);
        }
    }
}
```

```java
public void makeSet(String node) {
    parent.put(node, node)
    rank.put(node, 0);
}
```

#### 2.2. 크루스칼 알고리즘 (Kruskal's algorithm) 구현

```java
import java.util.Collections;

public ArrayList<Edge> kruskalFunc(ArrayList<String> vertex, ArrayList<Edge> edges) {
    ArrayList<Edge> mst = new ArrayList<Edge>();
    Edge currentEdge;
    
    // 초기화 진행
    for (int idx = 0; idx < vertex.size(); idx++) {
        makeSet(vertex.get(idx));
    }
    
    // 간선의 가중치를 기반으로 정렬함
    Collections.sort(edges);
    
    for (int idx = 0; idx < edges.size(); idx++) {
        currentEdge = edges.get(idx);
        if (find(currentEdge.nodeE) != find(currentEdge.nodeN)) {
            union(currentEdge.nodeE, currentEdge.nodeN);
            mst.add(currentEdge);
        }
    }
    return mst;
}
```

#### 2. 3. 전체 알고리즘 구현

```java
import java.util.Collections;
import java.util.HashMap;
import java.util.ArrayList;
import java.util.Arrays;

public class KruskalAlgo {
    HashMap<String, String> parent = new HashMap<String, String>();
    HashMap<String, Integer> rank = new HashMap<String, Integer>();
    
    public String find(String node) {
        // path compresion
        if (this.parent.get(node) != node) {
            this.parent.put(node, this.find(this.parent.get(node)));
        }
        return this.parent.get(node);
    }
    
    public void union(String nodeE, String nodeN) {
        String root_1 = this.find(nodeE);
        String root_2 = this.find(nodeN);
        
        // union-by-rank
        if (this.rank.get(root_1) > this.rank.get(root_2)) {
            this.parent.put(root_2, root_1);
        } else {
            this.parent.put(root_1, root_2);
            if (this.rank.get(root_1) == this.rank.get(root_2)) {
                this.rank.put(root_2, this.rank.get(root_2) + 1);
            }
        }
    }
    
    public void makeSet(String node) {
        this.parent.put(node, node);
        this.rank.put(node, 0);
    }
    
    public ArrayList<Edge> kruskalFunc(ArrayList<String> vertex, ArrayList<Edge> edges) {
        ArrayList<Edge> mst = new ArrayList<Edge>();
        Edge currentEdge;
        
        // 초기화 진행
        for (int idx = 0; idx < vertex.size(); idx++) {
            this.makeSet(vertex.get(idx));
        }
        
        // 간선의 가중치를 기반으로 정렬함
        Collections.sort(edges);
        
        for (int idx = 0; idx < edges.size(); idx++) {
            currentEdge = edges.get(idx);
            if (this.find(currentEdge.nodeE) != this.find(currentEdge.nodeN)) {
                this.union(currentEdge.nodeE, currentEdge.nodeN);
                mst.add(currentEdge);
            }
        }
        
        return mst;
    }
}
```

#### 3. 시간 복잡도
- 크루스칼 알고리즘의 시간 복잡도는 O(E log E)
  - 간선을 비용 기준으로 정렬하는 시간에 따라 결정됨
  1. 모든 정점을 각각 집합으로 만듬
  2. 모든 간선을 비용이 작은 간선부터 양 끝의 두 정점을 비교함
  3. 두 정점의 최상위 정점을 확인하고, 서로 다르면 두 정점을 연결함
      - 사이클이 생기지 위해서임
