##  14주차 백트래킹

#### 1.백 트래킹(backtracking)
- '되돌아 가다'의 의미로, 퇴각 검색 또는 퇴각 기법
- 제약 조건 만족 문제 (Constraint Satisfaction Problem) 에서 해를 찾기 위한 전략
  - 만족하는 해를 찾기 위해, 후보들의 제약 조건을 점진적으로 확인하고, 이를 만족할 수 없다고 판단되면 즉시 backtrack (해당 후보군은 다시 확인하지 않음)하고, 다른 후보를 검색하여 최적의 해를 찾는 방법
- 실제 구현시, 고려할 수 있는 모든 경우의 수 (후보군)를 상태공간트리로 확인함
  - 모든 후보들을 DFS 방식으로 탐색을 확인함
  - 상태 공간 트리를 검색하면서, 조건이 적절하지 않으면 다음 후보를 탐색함

#### 1.2. n-여왕(n-queen) 문제
- nxn 크기의 체스판에 n개의 여왕을 서로 공격할 수 없도록 배치하는 문제
- 여왕은 아래와 같이 이동할 수 있으므로, 여왕끼리는 공격할 수 없는 위치에 배치해야 함
- 대표적인 백트래킹 문제
![nn](img/img_14_02.png)

#### 1.3. n-여왕 문제에서pruning
- pruning (가지치기): 조건에 맞지 않으면 포기하고 다른 루트로 바로 돌아서서, 탐색의 시간을 절약하는 기법
- 각 행에는 하나의 퀸만 배치가 가능함 (퀸은 수평 이동이 불가능함)
- 첫 번째 행부터행부터 퀸을 배치하고, 그 다음 행에 첫 번째 행의 퀸이 이동할 수 없는 위치를 찾아 퀸을 배치함
- 전 행에 배치한 퀸때문에 다음 행에 배치할 퀸들이 이동할 수 없는 위치가 없을 경우는 더 이상 퀸을 배치하지 않고, 이전 행의 퀸의 배치를 바꿈
![nn](img/img_14_03.png)

#### 1.4. n-여왕 문제에서 promising
- Promising: 해당 루트가 조건에 맞는지를 검사하는 기법

- 해당 루트가 조건에 맞는지를 검사하는 기법을 활용하여,
- 현재까지 앞선 행에서 배치한 퀸이 이동할 수 없는 위치가 있는지를 다음과 같은 조건으로 확인
  - 한 행에 어차피 하나의 퀸만 배치 가능하므로 수평 체크는 별도로 필요하지 않음
![nn](img/img_14_04.png)

#### 1.5. n-Queen 구현

```java
import java.util.ArrayList;

public class n_Queen {
    
    public boolean isAvailable(ArrayList<Integer> candidate, Integer currentCol) {
        Integer currentRow = candidate.size();
        for (int idx = 0; idx < currentRow; idx++) {
            if ((candidate.get(idx) == currentCol) || (Math.abs(candidate.get(idx) - currentCol) == currentRow - idx) ) {
                return false;
            }
        }
        return true;
    }
    
    public void checkFunc(Integer n, Integer currentRow, ArrayList<Integer> currentCandidate) {
        if (currentRow == n) {
            System.out.println(currentCandidate);
            return ;
        }
        
        for (int idx = 0; idx < n; idx++) {
            if (this.isAvailable(currentCandidate, idx) == true) {
                currentCandidate.add(idx);
                this.checkFunc(n, currentRow + 1, currentCandidate);
                currentCandidate.remove(currentCandidate.size() - 1);
            }
        }
    }   
}
```
