#### 주피터 노트북 메뉴얼
- 셀에 코드 대신에 텍스트를 쓰고 싶으면, 셀의 타입을 Markdown 으로 변경하면 됨
   - 특정 셀의 코드가 실행이 완료되지 않으면, 다른 셀의 코드는 실행이 안됨
- 셀에 코드를 넣고 실행하면 결과가 나옴   
- 코드가 중지되지 않고 계속 실행 된다면, 상단 메뉴 중 네모(STOP) 버튼을 눌러, 셀을 중단할 수 있음

#### (1) 셸 선택 모드 (Command Mode)

<span style="color:blue">**파란색은 셸 선택 상태임 (선택상태)**</span> <br>
[ESC] 또는 [Ctrl] + [m]를 눌러 셀이 아래와 같이 파란색이 된 상태(셀 선택 모드)에서 해당 단축키 누름
![nn](img/img_1.png)
<span style="color:blue">**a**</span> : 위에 새로운 셀 추가 <br>
<span style="color:blue">**b**</span>: 아래에 새로운 셀 추가


<span style="color:blue">**c**</span> : 셀 복사하기<br>
<span style="color:blue">**v**</span> : 셀 붙여넣기<br>
<span style="color:blue">**x**</span> : 셀 잘라내기<br>
<span style="color:blue">**dd**</span> : 셀 삭제하기<br>
<span style="color:blue">**p**</span> : 셀 아래에 붙여넣기

 

<span style="color:blue">**o**</span> : 실행결과 열기/닫기<br>
<span style="color:blue">**m**</span> : Markdown으로 변경<br>
<span style="color:blue">**y**</span> : Code로 변경<br>
<span style="color:blue">**Shift + m**</span> : 선택 셀과 아래 셀과 합치기<br>
<span style="color:blue">**Ctrl + s 또는 s**</span>  : 파일 저장<br>
<span style="color:blue">**Enter**</span>  : 선택 셀의 코드 입력 모드로 돌아가기<br>

#### (2) 코드 입력 모드 (Edit code)
<span style="color:green">**초록색은 코드 입력 가능 상태 (입력상태)**</span> <br>
![nn](img/img_2.png)

<span style="color:green">**Ctrl + Enter**</span> : 입력 셀 실행 <br>
<span style="color:green">**Shift + Enter**</span> : 입력 셀 실행 후 아래 셀로 이동 (없으면 새로운 셀 추가)<br>
<span style="color:green">**Alt + Enter**</span> : 입력영역 실행 후 아래 새로운 영역 추가<br>

<span style="color:green">**Ctrl + a**</span> : 선택 셀의 코드 전체 선택<br>
<span style="color:green">**Ctrl + z**</span> : 선택 셀 내 실행 취소<br>
<span style="color:green">**Ctrl + y**</span> : 선택 셀 내 다시 실행<br>
<span style="color:green">**Ctrl + /**</span> : 커서 위치 라인 주석처리<br>

<span style="color:green">**Shitf + Ctrl + -**</span> : 커서 위치에서 셀 둘로 나누기<br>

#### (5) 목록(List)

 > 순서가 있는 목록은 다음과 같이 숫자로 작성이 가능함 
1. 1주차
2. 2주차
3. 3주차

> 순서가 없는 목록은 *, +, -로 사용이 가능하다. 결과는 모두 ·으로 표시됨
* 별표(*)
+ 플러스(+)
- 마이너스(-)

> 들여쓰기로 level을 표시할 수 도 있음
 - level은 띄어쓰기나 tab 후 내용을 작성하면 됨
 
* 1주차
 * 1 단계
 * 2 단계
   * 2-1 단계
   
1. 1 주차
 * level 1
 * level 2
2. 2 주차
 * level 1
     * level 1-1
