#### 2. JAVA 코드 테스트
> 주피터 노트북에서, 메뉴의 new를 클릭 후, Java를 선택하여 진행

#### 2.1 클래스 선언 및 실행
- main 메서드가 있더라도, 일반 클래스처럼 객체로 선언하고 진행

```java
class Program {
   public static void main(String[] args) {
       System.out.println("Hello, World!");
   }
   int TextMessage() {
       return 1 + 2;
   }
}
```

#### 객체 생성 후 실행

```java
Program Program = new Program();
```

```java
Program.TextMessage();
```

#### 메서드 실행 방법

```java
Program.main(new String[0]); // 클래스 메서드 실행, main 선언에서 인자를 넣어주므로, new String[0]으로 넣음
```

```java
new Program().TextMessage(); // 객체로 만들며 바로 실행하는 방법
```

#### 2.2 패키지 선언 및 사용 방법
- 주피터 노트북에서 패키지 임포트하면 해당 패키지를 사용하는 것이 가능함

```java
import java.util.Scanner;
```

```java
// 일부러, 2.1. 절의 클래스명과 동일하게 선언을 해보았음. 미리 선언된 클래스/변수도 언제든 다른 셀에서 새로운 코드로 변경 가능
class Program {
  public static void main(String[] args) {
    Scanner inSen = new Scanner(System.in);
    System.out.println("단어를 입력하고 엔터 버튼을 누르세요. 누르지 않으면 다른 셀도 실행이 안됩니다");

    String inputWord = inSen.nextLine();
    System.out.println("입력한 단어는 " + inputWord );
  }
}
```

```java
Program inputTest = new Program();
```

```java
inputTest.main(new String[0]);
```

- 임포트된 패키지는 다른 셀에서도 사용이 가능함

```java
Scanner inSen = new Scanner(System.in);
```

```java
Scanner inSen = new Scanner(System.in);
System.out.println("당신의 이름을 입력하세요. 그리고 엔터를 누르세요.");

String inputName = inSen.nextLine();
```

- 셀에서 맨 마지막 라인에 기재된 객체와 변수는 바로 해당 값을 출력
