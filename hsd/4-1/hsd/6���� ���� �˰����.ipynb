### 2. 버블 정렬 (bubble sort)
* 이웃한 두 요소의 대소 관계를 비교하고 필요에 따라 교환을 반복하는 알고리즘


#### 버블 정렬 확인해보기: https://visualgo.net/en/sorting

#### 2-1. 코드로 구현하기

- 세 개의 데이터가 존재할 때 버블 정렬 알고리즘으로 정렬하기

````java
import java.util.ArrayList;
import java.util.Collections;

ArrayList<Integer>dataList = new ArrayList<Integer>();
dataList.add(8);
dataList.add(1);
dataList.add(6);

for(int idx = 0; idx <dataList.size() -1; idx++) {
    if(dataList.get(idx) > dataList.get(idx + 1)) {
        Collections.swap(dataList, idx, idx + 1);
    }
}
System.out.println(dataList);
````

````java
import java.util.ArrayList;
import java.util.Collections;

public class BubbleSort {
    public ArrayList<Integer> sort(ArrayList<Integer> dataList) {
        for (int idx = 0; idx < dataList.size() - 1 ;idx++) {
            boolean swap =false;
        
        for (int idx2 = 0; idx2 < dataList.size() -1 -idx ; idx2++) {
            if (dataList.get(idx2) > dataList.get(idx2 + 1)) {
                    Collections.swap(dataList, idx2, idx2 + 1);
                    swap = true;
            }
        }
        if ( swap == false) {
            break;
        }
    }
    return dataList;
    }
}
````

### 3. 선택 정렬 (selection sort)
  - 가장 작은 요소를 맨 앞으로 이동하고, 두 번째 작은 요소는 맨 앞에서 두 번째로 이동하는 등의 작업을 반복하는 알고리즘


#### 선택 정렬 확인해보기: https://visualgo.net/en/sorting

#### 3-1. 선택 정렬 구현하기

> 배열에 데이터가 저장되어 있다고 가정을 해봄 

* 데이터가 두 개 일때
  - 예: dataArray = [8, 1]
    - dataArray[0] > dataArray[1] 이므로 dataArray[0] 값과 dataArray[1] 값을 교환함
    
* 데이터가 세 개 일때
  - 예: dataArray = [9, 1, 8]
    - 처음 실행하면, 1, 9, 8 이 됨
    - 두 번째 실행하면, 1, 8, 9 가 됨
    
* 데이터가 네 개 일때
  - 예: dataArray = [9, 4, 3, 1]
    - 처음 한번 실행하면, 1, 4, 3, 9 가 됨
    - 두 번째 실행하면, 1, 3, 4, 9 가 됨
    - 세 번째 실행하면, 아무런 변화가 없음

#### 3-2.  구현해보기
1. for (int std = 0; std < dataList.size() - 1; std++) 로 반복
2. low = std 로 설정
3. for (int idx = std + 1; idx < dataList.size(); idx++) 로, std 이후부터 반복을 진행함
   - 안의 반복문에서 dataList[low] > dataList[idx] 이면, 
     - low = num
4. dataList[low] 와 dataList[idx]를 스왑

````java
import java.util.ArrayList;
import java.util.Collections;

public class SelectionSort {
    public ArrayList<Integer> sort(ArrayList<Integer> dataList) {
        int low;
        for( int std = 0; std < dataList.size() - 1; std++) {
            low = std;
            for (int idx = std + 1; idx < dataList.size(); idx++) {
                if (dataList.get(low) > dataList.get(idx)) {
                    low = idx;
                }
            }
            Collections.swap(dataList, low, std);
        }
        return dataList;
    }
}
````

#### 테스트

````java
ArrayList<Integer> dataSet = new ArrayList<Integer>();

for (int i = 0; i < 100; i++) {
    dataSet.add((int)(Math.random() * 100));
}
SelectionSort sSort = new SelectionSort();
sSort.sort(dataSet);
````

#### 3-3. 알고리즘 분석
* 반복문(for 문)이 중복됨 O($n^2$)
  - 최악의 경우, <font size=5em>$\frac { n * (n - 1)}{ 2 }$</font>

### 4.삽입 정렬 (insertion sort) 

#### 4-1. 삽입 정렬 (insertion sort) 란?
- 삽입 정렬은 두 번째 인덱스부터 시작함
- 인덱스 앞에 있는 데이터부터 비교해서 값이 더 작으면, 값을 뒤의 인덱스로 복사함
* 값이 더 큰 데이터를 만날때까지 반복하고 큰 수를 만난 위치 바로 뒤에 값을 이동시킴

#### 삽입정렬 확인해보기: https://visualgo.net/en/sorting

#### 4-2. 구현해보기

* 데이터가 네 개 일때 
  - 예: dataList = [8, 4, 2, 5]
    - 처음은 key값은 8, 인덱스(0) - 1 은 0보다 작으므로 종료: [8, 4, 2, 5]
    - 두 번째는 key값은 4, 8보다 4가 작으므로 자리를 교환하고 종료: [4, 8, 2, 5]
    - 세 번째는 key값은 2, 8보다 2가 작으므로 자리를 교환하고, 다시 3보다 2가 작으므로 종료: [2, 4, 8, 5]
    - 네 번째는 key값은 5, 8보다 5가 작으므로 자리를 교환하고, 4보다는 5가 크므로 종료: [2, 4, 5, 9]     

#### 알고리즘 로직
1. for (int idx = 0; idx < dataList.size() - 1; idx++) 로 반복
2. 반복문 안에서, for (int idx2 = index + 1; idx2 > 0; idx2--) 으로 반복
   - 반복문 안에서 if (dataList.get(idx2) < dataList.get(idx2 - 1))되면 스왑을 진행함

````java
import java.util.ArrayList;
import java.util.Collections;

public class InsertionSort {
    public ArrayList<Integer> sort(ArrayList<Integer> dataList) {
        for (int idx = 0; idx < dataList.size() - 1; idx++) {
            for (int idx2 = idx + 1; idx2 > 0; idx2--) {
                if (dataList.get(idx2) < dataList.get(idx2 - 1)) {
                    Collections.swap(dataList, idx2, idx2 - 1);
                } else {
                    break;
                }
            }
        }
        return dataList;
    }
}
````

````java
ArrayList<Integer> dataSet = new ArrayList<Integer>();

for (int idx = 0; idx < 100; idx++) {
    dataSet.add((int)(Math.random() * 100));
}

InsertionSort iSort = new InsertionSort();
iSort.sort(dataSet);
````

#### 4-3. 알고리즘 분석
* 반복문(for 문)이 중복됨 O($n^2$)
  - 최악의 경우, <font size=5em>$\frac { n * (n - 1)}{ 2 }$</font>
* 완전 정렬이 되어 있다면 최선은 O(n)

### 5.퀵 정렬 (quick sort) 

* 기준점(pivot)을 정하고, 기준점보다 작은 데이터는 왼쪽(left)으로 큰 데이터는 오른쪽(right)으로 모으는 함수를 만듬
* 각 왼쪽(left), 오른쪽(right)은 재귀용법을 사용해서 다시 동일 함수를 호출하고 위의 작업을 반복함
* 함수는 왼쪽(left) + 기준점(pivot) + 오른쪽(right) 을 반환함
* <font color='#BF360C'>정렬 알고리즘에서 가장 중요함</font>

#### 5-1. 퀵정렬 구현해보기

* QuickSort.sort() 함수 만들기
  - 리스트 갯수가 한개라면 해당 리스트를 반환
  - 아닌 경우는 리스트 맨 앞의 데이터를 기준점(pivot)으로 배치
  - left, right 리스트 변수를 만들고,
  - 맨 앞의 데이터를 뺀 나머지 데이터를 기준점과 비교(pivot)함
    - 기준점보다 작으면 left.add
    - 기준점보다 크면 right.add
  - QuickSort.sort(left) + pivot + QuickSort.sort(right)을 방식으로 재귀함

````java
import java.util.ArrayList;
import java.util.Arrays;

public class QuickSort {
    public ArrayList<Integer> sort(ArrayList<Integer> dataList) {
        if (dataList.size() <= 1) {
            return dataList;
        }
        int pivot = dataList.get(0);
        
        ArrayList<Integer> leftArr = new ArrayList<Integer>();
        ArrayList<Integer> rightArr = new ArrayList<Integer>();
        
        for(int idx = 1 ; idx <dataList.size(); idx++) {
            if (dataList.get(idx) > pivot) {
                rightArr.add(dataList.get(idx));
            } else {
                leftArr.add(dataList.get(idx));
            }
        }
        
        ArrayList<Integer> mergedArr = new ArrayList<Integer>();
        mergedArr.addAll(this.sort(leftArr));
        mergedArr.addAll(Arrays.asList(pivot));
        mergedArr.addAll(this.sort(rightArr));
        
        return mergedArr;
                
    }
}
````

````java
ArrayList<Integer> dataSet = new ArrayList<Integer>();

for (int idx = 0; idx < 100; idx++) {
    dataSet.add((int)(Math.random() * 100));
}
QuickSort qSort = new QuickSort();
qSort.sort(dataSet);

````

#### 5-2. 알고리즘 분석
* 시간복잡도는 O(n log n)
  - 단, 최악의 경우 
    - 이미 정렬된 배열에서 pivot이 가장 크거나, 가장 작으면 가장 큰 시간이 소요됨
    - 모든 데이터를 비교할 수 있음
    - O($n^2$)
