#### 팩토리얼 분석
- 2~4의 수로 계산해보기
  * 2! = 2 X 1
  * 3! = 3 X 2 X 1 = 3 X 2!
  * 4! = 4 X 3 X 2 X 1 = 4 X 3!
  
- 다음과 같은 규칙을 찾을 수 있음  
   * n! = n X (n - 1)!
   
  1. 함수를 만듬
  2. 함수(n) 은 n > 0!(1) 이면 return n X 함수(n - 1)
  3. 함수(n) 은 n = 0!(1) 이면 return n
  
- 대입을 통해 검증해보기

    1. 2! 연산
     - 함수(2) 일때, 2 > 1 이므로 2 X 함수(1)
       - 함수(1) 은 1 이므로, return 2 X 1 = 2 연산이 정확함 
       
    2. 그 다음 3! 연산 
     - 함수(3) 이면, 3 > 1 이므로 3 X 함수(2)
       - 함수(2) 는 1번 연산에 의해 2! 이므로, return 2 X 1 = 2 
       - 3 X 함수(2) = 3 X 2 X 1 = 6 연산이 정확함
       
    3. 그 다음 4! 연산 
     - 함수(4) 이면, 4 > 1 이므로 4 X 함수(3)
       - 함수(3) 은 2번 연산에 의해 3 X 2 X 1 = 6 
       - 4 X 함수(3) = 4 X 3 X 2 X 1 = 24 연산이 정확함 

###  코드로 구현해보기

![nn](img/img_5_03.png)

````java
public class Factorial {
    public int factorialFunc(int n) {
        if(n > 1) {
            return n * this.factorialFunc(n -1);
        } else {
            return 1;
        }
    }
}
````

````java
Factorial funcObj = new Factorial();
funcObj.factorialFunc(4);
````

````java
Factorial funcObj = new Factorial();
funcObj.factorialFunc(3);
````

````java
Factorial funcObj = new Factorial();
funcObj.factorialFunc(2);
````

#### 2. 재귀의 연산 방법의 종류
![nn](img/img_5_04.png)

```java
// 연산의 방법 1
function(입력) {
    if (입력 <= 규정 값) {         // 입력이 규정 값보다 작으면
        return 값                  // 값을 반환하고 종료
    } 
    return function(입력 - 1);
}
```

```java
// 연산의 방법 2
function(입력) {
    if (입력 > 규정 값) {                     // 입력이 규정 값 이상이면
        return function(입력 - 1);           // 입력보다 작은 값
    } else {
        return 값; // 값을 반환하고 종료
    }
}
```

#### 코드로 구현해보기

````java
public class Factorial {
    public int factorialFunc(int n) {
        if(n <= 1) {
            return 1;
        } else {
            return n * this.factorialFunc(n-1);
        }
    }
}
````

````java
Factorial funcObj = new Factorial();
funcObj.factorialFunc(4);
````

````java
Factorial funcObj = new Factorial();
funcObj.factorialFunc(3);
````

````java
Factorial funcObj = new Factorial();
funcObj.factorialFunc(2);
````

### 알고리즘의 시간 복잡도와 공간 복잡도

- 시간 복잡도: 알고리즘 실행 속도
- 공간 복잡도: 알고리즘이 사용하는 메모리 공간

#### 알고리즘 표기법

- Big O (빅-오) 표기법 -> O(N)
  - 알고리즘 최악의 실행 시간을 표기함
  - **가장 보편적으로 사용함**
  - **어느 정도의 성능은 보장한다는 의미이기 때문**

- Ω (오메가) 표기법 ->  Ω(N)
  - 최상의 실행 시간을 표기함

- Θ (세타) 표기법 -> Θ(N)
  - 평균 실행 시간을 표기함

> 최상, 평균, 최악 중, 최악의 시간인 Big-O를 익히면 됨 

#### 3. Big-O 표기법

* O(입력)
  - 입력되는 n 에 따라 결정되는 시간 복잡도 함수
  - O(1), O($log n$), O(n), O(n$log n$), O($n^2$), O($2^n$), O(n!)등으로 표기함
  - 입력 n 의 크기에 따라 기하급수적으로 시간 복잡도가 늘어날 수 있음
    - O(1) < O($log n$) < O(n) < O(n$log n$) < O($n^2$) < O($2^n$) < O(n!)
      
#### Big-O 표기 방법
  - 시간 복잡도 함수가 3$n^2$ + 4n 이라면
      - 가장 높은 차수는 3$n^2$ 
      - 상수는 실제 큰 영향이 없음 
      - 결국 빅 오 표기법으로는 O($n^2$)
      
#### 재귀 알고리즘의 시간 복잡도 및 공간 복잡도

* factorial(n) 은 n - 1 번의 factorial() 함수를 호출해서, 곱셈을 함 
  - n-1번 반복문을 호출한 것과 동일
  - factorial() 함수를 호출할 때마다, 지역변수 n 이 생성됨

* 시간 복잡도/공간 복잡도는 O(n-1) 이므로 결국, 둘 다 O(n)
