## 12 주차 그리디 알고리즘

### 1. 그리디 알고리즘(greedy algorithm)
-  탐욕 알고리즘 또는 greedy algorithm 이라고 함
- 최적의 해에 가까운 값을 구하기 위해 사용됨
- 여러 경우의 수 중에 하나를 결정할 때, 최적에 근사하다고 보는 경우의 수를 선택하는 방법으로, 최종적인 값을 구함

#### 1.1. 최소 개수 동전 교환 문제
  - 10,000원으로 가격이 3,330원어치 음료수를 구입하고 최소의 동전 개수를 반환하려면? 잔돈은 6,670원
    (동전의 유형은 500원, 100원, 50원, 10원, 1원)
    - 액수가 가장 큰 동전부터 최대한으로 반환하는 방식으로 값을 채워 구현이 가능함
    - 탐욕 알고리즘으로 최적이라고 생각되는 경우의 수를 선택함

```java
public class greedyAlgo {
    public void coinFunction(Integer price, ArrayList<Integer> coinList) {
        Integer sumCoin = 0;
        Integer numCoin = 0;
        ArrayList<Integer> checking = new ArrayList<Integer>();
        
        for (int idx = 0; idx < coinList.size(); idx++) {
            numCoin = price / coinList.get(idx);
            sumCoin += numCoin;
            price -= numCoin * coinList.get(idx);
            checking.add(numCoin);
            System.out.println(coinList.get(idx) + "원: " + numCoin + "개");
        }
        System.out.println("동전의 갯수: " + sumCoin);
    }
}
```

```java
greedyAlgo gObject = new greedyAlgo();
ArrayList<Integer> coinList = new ArrayList<Integer>(Arrays.asList(500, 100, 50, 10, 1));
gObject.coinFunction(6670, coinList);
```

#### Comparable 인터페이스
- Comparable 와 Comparator는 정렬을 구현하기 위해 사용됨 

```java
// 10주차 그래프 edge 구현
public class Edge implements Comparable<Edge> { //정렬할 객체에 implements 로 Comparable 인터페이스를 추가하여 구현함
    public int weight;
    public String node;
    
    public Edge (int weight, String node) {
        this.weight = weight;
        this.node = node;
    }
    
    @Override
    public int compareTo(Edge edge) { // Comparable은 compareTo() 메서드를 override 해서 구현함
        return this.weight - edge.weight;
    }
}

Edge edge1 = new Edge(15, "A");
Edge edge2 = new Edge(10, "A");
Edge edge3 = new Edge(13, "A");
Edge edge4 = new Edge(16, "A");
Edge[] edges = new Edge[]{edge1, edge2, edge3, edge4};
Arrays.sort(edges);
for (int idx = 0; idx < edges.length; idx++) {
    System.out.println(edges[idx].weight);
}
```

#### Comparator 인터페이스

```java
public class Edge implements Comparable<Edge> {
    public int weight;
    public String node;
    
    public Edge (int weight, String node) {
        this.weight = weight;
        this.node = node;
    }
    
    @Override
    public int compareTo(Edge edge) {
        return this.weight - edge.weight;
    }
}
Edge edge1 = new Edge(15, "A");
Edge edge2 = new Edge(10, "A");
Edge edge3 = new Edge(13, "A");
Edge edge4 = new Edge(16, "A");
Edge[] edges = new Edge[]{edge1, edge2, edge3, edge4};
Arrays.sort(edges, new Comparator<Edge>() { //Arrays.sort() 메서드는 인자를 두 개 받고, 두 번째에 Comparator 클래스를 넣음
    @Override
    public int compare(Edge objectI_1, Edge objectI_2) { //compare() 메서드를 override 해서 구현
        return objectI_2.weight - objectI_1.weight;
    } //Edge 클래스에 Comparable이 정의가 되었지만, Comparator 클래스를 기준으로 정렬이 됨
});

for (int idx = 0; idx < edges.length; idx++) {
    System.out.println(edges[idx].weight);
}
```

#### 1.3. 부분 배낭(fractional knapsack) 문제
  - n개의 물건이 있고, 배낭이 한정된 무게의 물건들을 담을 수 있을 때(c), 배낭에 넣을 물건들이 최대의 가치를 가지도록 하는 문제
    - 각 물건은 무게(w)와 가치(v)로 표현될 수 있음
    - 물건은 부분적으로 배낭에 넣어질 수 있음, 그래서 Fractional Knapsack으로 부름
 ![nn](img/img_11_01.png)

```java
//2차원 배열로 구현
Integer[][] objList = { {10, 60000}, {15, 50000}, {25, 4000}, {50, 1000} };
```

```java
objList.length
```

```java
public class GreedyAlgo {
    public void knapsackFunction(Integer[][] objectList, double capacity) {
        double sumValue = 0.0;
        double fraction = 0.0;
        
        Arrays.sort(objectList, new Comparator<Integer[]>() {
            @Override
            public int compare(Integer[] objectI_1, Integer[] objectI_2) {
                return (objectI_2[1] / objectI_2[0]) - (objectI_1[1] / objectI_1[0]);
            }
        });
                    
        for (int idx = 0; idx < objList.length; idx++) {
            if ( (capacity - (double)objList[idx][0]) > 0 ) {
                capacity -= (double)objList[idx][0];
                sumValue += (double)objList[idx][1];
                System.out.println("무게:" + objList[idx][0] + ", 가치:" + objList[idx][1]);
            } else {
                fraction = capacity / (double)objList[idx][0];
                sumValue += (double)objList[idx][1] * fraction;
                System.out.println("무게:" + objList[idx][0] + ", 가치:" + objList[idx][1] + ", 비율:" + fraction);                
                break;
            }
        }
        System.out.println("담을 수 있는 가치:" + sumValue + "원");
    }
}
```

```java
GreedyAlgo grObject = new GreedyAlgo();
Integer[][] orbjectList = { {10, 60000}, {15, 50000}, {25, 4000}, {50, 1000} };
gObject.knapsackFunction(objectList, 40.0);
```
