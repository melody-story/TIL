## 12주차 힙

#### 1. 힙(heap) 이란?
- 완전 이진 트리의 일종으로 최대값과 최소값을 빠르게 찾기 위해 고안됨
  - 완전 이진 트리: 노드를 삽입할 때 최하단 왼쪽 노드부터 차례대로 삽입하는 트리

#### 1.2. 힙의 구조
- 최대 힙(max Heap): 모든 노드의 키 값이 자식 노드의 키 값보다 작지 않은 완전 이진 트리
- 최소 힙(min Heap): 모든 노드의 키 값이 자식 노드의 키 값보다 크지 않은 힙 트리

- 힙은 다음과 같은 조건을 가지고 있음
  * 최대 힙일 때, 각 노드의 값은 해당 노드의 자식 노드가 가진 값보다 크거나 같음
  * 최소 힙일 때, 각 노드의 값은 해당 노드의 자식 노드가 가진 값보다 크거나 작음

#### 최대 힙 구현 
- 인덱스 번호는 1번부터 시작하도록 변경

![nn](img/img_12_02.png)

````java
import java.util.ArrayList;

public class Heap {
    public ArrayList<Integer> heapArr = null;
    
    public Heap (Integer data) {
        heapArr = new ArrayList<Integer>();
        
        heapArr.add(null);
        heapArr.add(data);
    }
    
    public boolean insert(Integer data) {
        if (heapArr == null) {
            heapArr = new ArrayList<Integer>();
            
            heapArr.add(null);
            heapArr.add(data);
        } else {
            heapArr.add(data);
        }
        return true;
    }    
}
````

````java
Heap heapTest = new Heap(15);
heapTest.insert(12);
heapTest.insert(7);
heapTest.insert(10);
heapTest.insert(8);
heapTest.insert(5);

System.out.println(heapTest.heapArr);
````

#### 특정 노드 위치 알아내기
![nn](img/img_12_04.png)

  - 부모 노드 인덱스 번호 = 자식 노드 인덱스 번호 / 2
  - 왼쪽 자식 노드 인덱스 번호  = 부모 노드 인덱스 번호 * 2
  - 오른쪽 자식 노드 인덱스 번호 = 부모 노드 인덱스 번호 * 2 + 1

#### Collections.swap() 메서드 이해하기
- 두개의 데이터의 위치를 서로 바꾸는 것임
    
```
import  java.util.Collections;

Collections.swap(List arr, int x, int y)
```
- arr : 데이터가 저장된 배열
- x : 스왑할 인덱스의 번호
- y : 스왑할 인덱스의 번호

````java
import java.util.Collections;

ArrayList<Integer> heapArr = new ArrayList<Integer>();
heapArr.add(1);
heapArr.add(2);
System.out.println(heapArr);

Collections.swap(heapArr, 0, 1);
System.out.println(heapArr);
````

#### 힙에서 노드 삽입 및 변경 구현

````java
import java.util.ArrayList;

public class Heap {
    public ArrayList<Integer> heapArr = null;
    
    public Heap (Integer data) {
        heapArr = new ArrayList<Integer>();
        
        heapArr.add(null);
        heapArr.add(data);
    }
    
    public boolean transUp(Integer inserted_idx) {
        if (inserted_idx <= 1) {
            return false;
        }
        Integer parent_idx = inserted_idx / 2;
        if(this.heapArr.get(inserted_idx) > this.heapArr.get(parent_idx)) {
            return true;
        } else {
            return false;
        }
    }
    
    public boolean insert(Integer data) {
        Integer inserted_idx, parent_idx;
        
        if (heapArr == null) {
            heapArr = new ArrayList<Integer>();
            
            heapArr.add(null);
            heapArr.add(data);
            return true;
        }
       
        this.heapArr.add(data);
        inserted_idx = this.heapArr.size() - 1;
        
        while(this.transUp(inserted_idx)) {
        
            parent_idx = inserted_idx / 2;
            Collections.swap(this.heapArr, inserted_idx, parent_idx);
            inserted_idx = parent_idx;
        }
        return true;   
          
    }
    
    
}
````

#### 1.5. 힙의 시간 복잡도
  - 트리의 높이가 h이면, n개의 노드를 가지는 heap에 데이터 삽입 또는 삭제시, 최악의 경우 루트 노드부터 터미널 노드까지 비교해야 함
  - $h = log_2{n} $ 에 근접하므로, 시간 복잡도는 $ O(log{n}) $
