#### 1. JRE 와 JDK
- JRE(Java Runtime Enviroment) : 자바로 만들어진 프로그램을 실행시키는데 필요한 환경을 제공해주는 프로그램
- JDK(Java Development Kit) : 자바 프로그래밍 작성시 필요한 라이브러리,컴파일러 및 패키지 등을 포함하고 있음

> 따라서, JAVA 프로그램을 짜려면 JDK를 설치해야 하며, JDK 를 설치하면, 별도로 JRE 를 설치할 필요는 없음

#### 1.2. Java SE, Java EE, Java ME
- SE 는 Software Edition 의 약자로 개발자들을 위한 에디션
- EE 는 Enterprse Edition 의 약자로 기업용 JAVA 소프트웨어 개발을 위한 라이브러리들이 추가된 에디션
- ME 는 Micro Edition 의 약자로 임베디드용 JAVA 소프트웨어 개발을 위한 에디션

> Java SE 의 JDK 를 다운로드 받으면 됨

#### 1.3. JDK 다운로드 및 설치

1. JDK를 다운로드 받는 사이트
   - 다운로드 페이지 (JDK Download): https://www.oracle.com/java/technologies/downloads/
   
![nn](img/img_4.png)

   - JAVA 다운로드 페이지의 인터페이스는 수시로 바뀔 수 있으므로, 위 내용을 참고로 검색을 통해 설치 방법을 참고하도록 함
<br><br>

2. 다운로드 받은 파일을 실행함
 - 설치 경로를 변경할 수는 있지만, 보통은 기본 경로를 사용하여 설치함
 
![nn](img/img_5.png)
 
3. PATH 환경 변수 설정이 필요함(windows)
   - JDK 설치 경로를 확인
       - x64 기본 설치 경로 : C:\Program Files\Java\jdk-[version]
       - x32 기본 설치 경로 : C:\Program Files (x86)\Java\jdk-[version]
   - 제어판 -> 고급 시스템 설정 -> 환경 변수 탭에서 PATH 를 찾아서 편집을 누르고, JDK 설치 경로의 하부 폴더인 bin 폴더까지를 경로로 추가
     - 예: C:\Program Files\Java\jdk-15.0.2.jdk\bin 
<br><br>

4. 설치한 JDK의 환경 변수 설정을 위해 "시스템 환경 변수 편집" 메뉴 실행
 - 시스템 속성에서 "환경 변수"를 클릭
![nn](img/img_6.png)
<br>
5. 시스템 변수에서 변수 이름을 “JAVA_HOME”으로 지정하고 JDK가 설치된 폴더를 지정
 - 변수 이름: JAVA_HOME
 - 변수 값: C:\Program Files\Java\jdk-18.0.2.1
 ![nn](img/img_7.png)
 <br>
6. 시스템 변수에서 변수 이름을 “CLASSPATH”으로 지정하고 “%JAVA_HOME%\lib”
변수 이름: CLASSPATH
변수 값: “%JAVA_HOME%\lib
![nn](img/img_8.png)<br>

7. path 변수에서 bin 폴더 경로를 추가<br>
변수 이름: CLASSPATH <br>
변수 값: C:\Program Files\Java\jdk-18.0.2.1<br>
<br>
8. 명령어를 실행하여 설치한 java와 javac(컴파일러) 버전을 확인
![nn](img/img_9.png)

#### 1.4. IntelliJ IDEA 설치
- 다음 링크를 통하거나 구글에서 intelliJ IDEA 로 검색을 통해 사이트 접속<br>
  - https://www.jetbrains.com/ko-kr/idea/download/
- Community 버전으로 다운로드 및 설치 <br>

![nn](img/img_10.png)
<br>
  - 설치 과정에서 반드시 해야하는 설정은 없으며, 윈도우의 경우 다음 네가지 옵션을 보통 선택함
    - Create Desktop Shortcut : 바로가기 생성 옵션
    - Update PATH variable(리부팅 필요) : 윈도우 환경변수에 IntelliJ IDEA 폴더 추가 옵션
    - Update context menu : 프로젝트로 폴더 메뉴 추가
    - Create Association : 디폴트로 .java 파일을 IntelliJ IDEA 로 오픈할 것인지에 대한 설정 (보통 .java만 선택)
  - 설치 후, 최초 실행시 팝업창에서 Do not import settings 선택 후, 오픈
  - Ultimate 버전은 유료 (학생용 라이센스를 받을 수 있음)<br>
  
![nn](img/img_11.png)

#### 5. IntelliJ IDEA 에서 JAVA 코드 실행하기
- IntelliJ IDEA 실행 후, New Project 선택
- Project SDK 에 설치한 JDK 버전이 정확한지 확인
- Next 누르고, 별도 체크 없이, Next 를 다시 한번 누른 후, 
  - Project location을 설정하기
  - 프로젝트 이름 설정하기<br>
  
![nn](img/img_13.png)

- 왼쪽 src 폴더에서 오른쪽 클릭 후, New -> JAVA class 선택 후, JAVA 코드 파일명 입력

![nn](img/img_14.png)
- 해당 파일에서 다음과 같이 작성
  - main 함수 작성시 main 만 키보드로 입력 후, 탭(Tab) 을 누르면, main 함수가 자동으로 작성됨
  - 그외에도 기본적인 함수의 자동 완성 기능을 지원함
  
  ```java
  public class HelloWorld {
        public static void main(String[] args) {
            System.out.println("Hello IntelliJ IDEA");
        }
  }
  ```
  
![nn](img/img_15.png)

-  마우스 오른쪽 클릭 후, Run 메뉴를 누르면 실행됨
  - Debug 메뉴를 통해 디버깅도 가능함
  
 ![nn](img/img_16.png) 
  
-  main() 함수부터 실행하므로, 주피터 노트북에서 코드 작성 후 기본적으로 main 함수 내에서 실행하도록 하면 됨
