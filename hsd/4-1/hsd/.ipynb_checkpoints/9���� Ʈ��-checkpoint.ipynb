## 9주차  트리(tree)

### 1. 트리
- 노드들 간에 계층적인 구조를 나타내는 자료구조
- 트리 중 이진 트리 (binary Tree) 형태의 구조로, 검색 알고리즘 구현을 위해 많이 사용됨

#### 트리 관련 용어
- 노드(node): 데이터를 저장하는 구성 요소 
- 루트(root): 부모가 없는 노드(A), 트리의 가장 윗부분에 위치하는 노드
- 부모 노드(parent node): 노드에서 다음 레벨에 연결된 노드
- 자식 노드(child node): 노드에서 상위 레벨에 연결된 노드
- 레벨(level): 최상위 노드를 Level 0으로 하였을 때, 하위 Branch로 연결된 노드의 깊이를 나타냄
- 터미널 노드(terminal node): 자식 노드가 하나도 없는 노드
- 형제 노드 (sibling node): 같은 부모 노드를 가진 노드

![nn](img/img_9_01.png)

#### 이진 트리와 이진 탐색 트리 (Binary Search Tree)
- 이진 트리: 노드의 최대 Branch가 2인 트리
- 이진 탐색 트리 (Binary Search Tree, BST): 이진 트리에 다음과 같은 추가적인 조건이 있는 트리
  - 왼쪽 노드는 해당 노드보다 작은 값, 오른쪽 노드는 해당 노드보다 큰 값을 가짐

### 2. 트리 구현 

#### 2.1. 노드 구현

```java
public class nodeMng {
    public class Node {
        Node left;
        Node right;
        int value;
        public Node (int data) {
            this.value = data;
            this.left = null;
            this.right = null;
        }
    }
}
```

#### 2.2. 이진 탐색 트리에 데이터 넣기

```java
public class nodeMng {
    Node head = null;
    public class Node {
        Node left;
        Node right;
        int value;
        public Node (int data) {
            this.value = data;
            this.left = null;
            this.right = null;
        }
    }
    
    public boolean inputNode (int data) {
        // 1번 경우: 노드가 전혀 없을 때
        if (this.head == null) {
            this.head = new Node(data);
        } else {
            //2번 경우 : 노드가 하나 이상 있을 때
            Node findNode = this.head;
            while(true) {
                // 2-1: 노드가 왼쪽에 들어가야할 때
               if (data < findNode.value) {
                    if (findNode.left != null) {
                        findNode = findNode.left;
                    } else {
                        findNode.left = new Node(data);
                        break;
                }
                // 2-2: 노드가 오른쪽에 들어가야할 때
            } else {
                if (findNode.right != null) {
                    findNode = findNode.right;
                } else {
                    findNode.right = new Node(data);
                    break;
                }
            }
        }
    }
    return true;
    }
}
```

```java
nodeMng bTree = new nodeMng();
bTree.inputNode(1);
bTree.inputNode(4);
bTree.inputNode(5);
bTree.inputNode(7);
```

#### 2.3. 이진 탐색 트리의 탐색 구현

```java
    public class Node {
        Node left;
        Node right;
        int value;
        public Node (int data) {
            this.value = data;
            this.left = null;
            this.right = null;
        }
}
  

public class nodeMng {
    Node head = null;

    public boolean inputNode(int data) {
        // 1번 경우: 노드가 전혀 없을 때
        if (this.head == null) {
            this.head = new Node(data);
        } else {
            // 2번 경우: 노드가 존재할 때
            Node findNode = this.head;
            while (true) {
                // 2-1: 노드의 왼쪽에 들어가야할 때
                if (data < findNode.value) {
                    if (findNode.left != null) {
                        findNode = findNode.left;
                    } else {
                        findNode.left = new Node(data);
                        break;
                    }
                // 2-2: 노드의 오른쪽에 들어가야할 때                    
                } else {
                    if (findNode.right != null) {
                        findNode = findNode.right;
                    } else {
                        findNode.right = new Node(data);
                        break;
                    }
                }
            }
        }
        return true;        
    }
    
    public Node check(int data) {
        // 1번 경우: 노드가 하나도 없을 때
        if (this.head == null) {
            return null;
        // 2번 경우: 노드가 존재할 때            
        } else {
            Node findNode = this.head;
            while (findNode != null) {
                if (findNode.value == data) {
                    return findNode;
                } else if (data < findNode.value) {
                    findNode = findNode.left;
                } else {
                    findNode = findNode.right;
                }
            }
            return null;
        }
    }
}
```

```java
nodeMng bTree = new nodeMng();
bTree.inputNode(1);
bTree.inputNode(4);
bTree.inputNode(5);
bTree.inputNode(7);
```

```java
Node testNode = bTree.check(4);
```

```java
testNode.right.value;
```

#### 2.4. 시간 복잡도
  - depth (트리의 높이) 를 h라고 표기한다면, O(h)
  - n개의 노드를 가진다면, $h = log_2{n} $ 에 가까우므로, 시간 복잡도는 $ O(log{n}) $
