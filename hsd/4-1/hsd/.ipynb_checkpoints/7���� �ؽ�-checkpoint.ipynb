## 6주차 해시(hash)

### 1. 해시 테이블
  - 키(key)를 통해 데이터를 받아올 수 있으므로 속도가 빠름
  - 키(Key)에 데이터(Value)를 매핑할 수 있는 구조
  - 해시 함수를 통해, 배열에 키에 대한 데이터를 저장할 수 있는 주소를 연산  

### 용어 정리
* 해시 함수(hash Function): 데이터를 고정된 길이의 값으로 리턴해주는 함수
* 해시 값(Hash Value): 해시 함수가 계산한 값
* 해시 테이블(hash Table): 키 값의 연산에 의해 직접 접근이 가능한 데이터 구조
* 슬롯(slot): 한 개의 데이터를 저장할 수 있는 공간

### 2. 해시 구현하기

#### 단순한 hash table 구현

```java
public class createHash {
    public Slot[] hashTable;
    
    public createHash(Integer size) {
        this.hashTable = new Slot[size];
    }
    
    public class Slot {
        String value;
        
        Slot(String value) {
            this.value = value;
        }
    }
}
```

#### 2-2. 해시 테이블에 해시 함수 추가

```java
public class createHash {
    public Slot[] hashTable;
    
    public createHash(Integer size) {
        this.hashTable = new Slot[size];
    }
    
    public class Slot {
        String value;
        
        Slot(String value) {
            this.value = value;
        }
    }
    
    public int hashFunction(String key) {
        return (int)(key.charAt(0) % this.hashTable.length);
    }
}
```

#### 2-3. 해쉬 테이블에 데이터를 저장하는 메서드 구현

- 각 배열은 객체의 주소를 담을 수 있는 공간만 할당

```java
public class Slot {
    String value;
    Slot(String value) {
        this.value = value;
    }
}

Slot[] hashTable = new Slot[10];
System.out.println(hashTable[0]);
```

```java
public class Slot {
    String value;
    Slot(String value) {
        this.value = value;
    }
}

Slot[] hashTable = new Slot[10];
hashTable[0] = new Slot("AAA");
System.out.println(hashTable[0]);
System.out.println(hashTable[0].value);
```

#### 2-4. 해쉬 테이블에 저장 기능 구현

```java
public class createHash {
    public Slot[] hashTable;
    
    public createHash(Integer size) {
        this.hashTable = new Slot[size];
    }
    
    public class Slot {
        String value;
        Slot(String value) {
            this.value = value;
        }
    }
    
    public int hashFunction(String key) {
        return (int)(key.charAt(0)) % this.hashTable.length;
    }
    
    public boolean saveData(String key, String value) {
        Integer address = this.hashFunction(key); 
        if (this.hashTable[address] != null) {
            this.hashTable[address].value = value;
        } else {
            this.hashTable[address] = new Slot(value);
        }
        return true;
    }
    
     
}
```

#### 2.5. 해쉬 테이블에 key 에 대한 데이터를 가져오는 기능 구현

```java
public class createHash {
    public Slot[] hashTable;
    
    public createHash(Integer size) {
        this.hashTable = new Slot[size];
    }
    
    public class Slot {
        String value;
        Slot(String value) {
            this.value = value;
        }
    }
    
    public int hashFunction(String key) {
        return (int)(key.charAt(0)) % this.hashTable.length;
    }
    
    public boolean saveData(String key, String value) {
        Integer address = this.hashFunction(key);
        if (this.hashTable[address] != null) {
            this.hashTable[address].value = value;
        } else {
            this.hashTable[address] = new Slot(value);
        }
        return true;
    }
    
    public String loadData(String key) {
        Integer address = this.hashFunction(key);
        if (this.hashTable[address] != null) {
            return this.hashTable[address].value;
        } else {
            return null;
        }
    }
    
    
}
```

```java
createHash mobject = new createHash(10);
mobject.saveData("AAA", "1111");
mobject.saveData("BBB", "2222");
mobject.loadData("AAA");
```

#### 3. 시간 복잡도
- 일반적인 경우(Collision이 없는 경우)는 O(1)
- 최악의 경우(Collision이 모두 발생하는 경우)는 O(n)
- Linear Probing, Chaining 기법 둘 다 동일
