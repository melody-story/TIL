#### 4주차 연결 리스트

#### 1. 연결 리스트  구조
* 링크드 리스트(Linked List)라고도 함
* 배열은 데이터가 연속하는 메모리 공간에 저장되어 있음
* 연결 리스트는 다음 데이터에 대한 정보를 화살표로 연결해서 관리하는 데이터 구조

* 연결 리스트 구조와 용어
  - 노드(node): 데이터값, 포인터로 구성됨
  - 포인터(pointer): 각 노드에서 연결한 노드의 정보를 가지고 있는 공간

<br>
* 연결 리스트의 형태

![nn](img/img_4_01.png)

#### 연결 리스트의 장단점
* 장점
  - 미리 데이터 공간을 미리 할당하지 않아도 됨
* 단점
  - 연결을 위한 별도의 데이터 공간이 필요함
  - 연결 정보를 찾는 시간이 필요하므로 접근 속도가 느림
  - 중간 데이터 삭제시, 앞뒤 데이터의 연결을 재구성해야 하는 부가적인 작업 필요

#### 1-1. 연결 리스트 구현하기

#### - 노드(node) 클래스 구현

```java
public class Node<E> {
    E data;
    Node<E> next = null;
    
    public Node(E data) {
        this.data = data;
    }
}
```

#### 노드와 노드간 연결해보기

```java
Node<Integer> node1 = new Node<Integer>(1);
Node<Integer> node2 = new Node<Integer>(2);
Node<Integer> node3 = new Node<Integer>(3);
```

```java
node1.next = node2;
node2.next = node3;
Node<Integer> head = node1; 
```

#### - 연결 리스트에 데이터 추가하기

```java
public class SimpleLinkedList<E> {
    public Node<E> head = null; 
    
    public class Node<E> {
        E data;
        Node<E> next = null;
        
        public Node(E data) {
            this.data = data;
        }
    }
    
    public void addNode(E data) {
        if (head == null) {
            head = new Node<E>(data);
        } else {
            Node<E> node = this.head;
            while (node.next != null) {
                node = node.next;
            }
            node.next = new Node<E>(data);
        }
    }
}
```

```java
SimpleLinkedList<Integer> ImLinkedList = new SimpleLinkedList<Integer>();
ImLinkedList.addNode(1);
ImLinkedList.addNode(2);
```

```java
ImLinkedList.head.next.data;
```

#### - 연결 리스트에 속해 있는 노드의 데이터가 출력되게 구현해보기

```java
public class SimpleLinkedList<E> {
    public Node<E> head = null; 
    
    public class Node<E> {
        E data;
        Node<E> next = null;
        
        public Node(E data) {
            this.data = data;
        }
    }
    
    public void addNode(E data) {
        if (head == null) {
            head = new Node<E>(data);
        } else {
            Node<E> node = this.head;
            while (node.next != null) {
                node = node.next;
            }
            node.next = new Node<E>(data);
        }
    }
    
    public void outputData() {
        if (head != null) {
            Node<E> node = this.head;
            System.out.println(node.data);
            while(node.next != null) {
                node = node.next;
                System.out.println(node.data);
            }
        }
    }
}
```

```java
SimpleLinkedList<Integer>ImLinkedList = new SimpleLinkedList<Integer>();
ImLinkedList.addNode(1);
ImLinkedList.addNode(2);
ImLinkedList.addNode(3);
ImLinkedList.addNode(4);

ImLinkedList.outputData();
```

#### - 연결리스트에서 노드를 삭제 하는 것 구현하기
![nn](img/img_4_02.png)

```java
public class SimpleLinkedList<E> {
    public Node<E> head = null; 
    
    public class Node<E> {
        E data;
        Node<E> next = null;
        
        public Node(E data) {
            this.data = data;
        }
    }
    
    public void addNode(E data) {
        if (head == null) {
            head = new Node<E>(data);
        } else {
            Node<E> node = this.head;
            while (node.next != null) {
                node = node.next;
            }
            node.next = new Node<E>(data);
        }
    }
    
    public void outputData() {
        if (head != null) {
            Node<E> node = this.head;
            System.out.println(node.data);
            while(node.next != null) {
                node = node.next;
                System.out.println(node.data);
            }
        }
    }
    
    public boolean deleteNode(E priData) {
        if (this.head == null) {
            return false;
        } else {
            Node<E> node = this.head;
            if (node.data == priData) {
                this.head = this.head.next;
                return true;
            } else {
                while(node.next != null) {
                    if (node.next.data == priData) {
                        node.next = node.next.next;
                        return true;
                    }
                    node = node.next;
                }
                return false;
            }
        }
    }
}
```

### 1-2. 이중 연결 리스트
* 더블 링크드 리스트(Double linked list)라고도 함  
* 양방향으로 연결되어 있어서 노드 탐색이 양쪽으로 모두 가능
  
![nn](img/img_4_03.png)

```java
public class DoubleLinkedList<E> {
    public Node<E> head = null;
    public Node<E> tail = null;
    
    public class Node<E> {
        E data;
        Node<E> prev = null;
        Node<E> next = null;
        
        public Node(E data) {
            this.data = data;
        }
    }
    
    public void addNode(E data) {
        if (this.head == null) {
            this.head = new Node<E>(data);
            this.tail = this.head;
        } else {
            Node<E> node = this.head;
            while (node.next != null) {
                node = node.next;
            }
            node.next = new Node<E>(data);
            node.next.prev = node;
            this.tail = node.next;
        }
    }
}
```
